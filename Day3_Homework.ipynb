```python
# Extract text from the homework PDF
from pypdf import PdfReader
import os
pdf_path = 'Day-03/Day3_Homework.pdf'
if not os.path.exists(pdf_path):
    raise FileNotFoundError(f'PDF not found at {pdf_path}. In Colab upload the PDF to this path or update `pdf_path`.')
reader = PdfReader(pdf_path)
lines = []
for p, page in enumerate(reader.pages):
    text = page.extract_text()
    if text:
        lines.extend(text.splitlines())
homework_text = '
'.join(lines)
# Save a plain-text copy for convenience
with open('Day-03/Day3_Homework_text.txt', 'w', encoding='utf-8') as f:
    f.write(homework_text)
# Print a preview
print('--- Homework PDF text preview (first 1000 chars) ---')
print(homework_text[:1000])
print('
--- End preview ---')
```

```python
# Parse likely task lines from the extracted text
import re, json
# heuristic: look for lines starting with 'Task' or a numbered list like '1.' or '1)'
raw = homework_text
pattern = re.compile(r'^\s*(?:Task\s++|\d+\.|\d+\))\s*(.*)', re.IGNORECASE|re.MULTILINE)
matches = pattern.findall(raw)
if matches:
    tasks = [m.strip() for m in matches]
else:
    # fallback: split by blank lines and take top paragraphs as tasks
    paragraphs = [p.strip() for p in raw.split('

') if p.strip()]
    tasks = paragraphs[:6]  # keep up to 6 items by default
print(f'Detected {len(tasks)} task(s):')
for i, t in enumerate(tasks, 1):
    print(f'  Task {i}:', t[:200])
with open('Day-03/homework_tasks.json','w',encoding='utf-8') as f:
    json.dump(tasks, f, indent=2, ensure_ascii=False)
```
